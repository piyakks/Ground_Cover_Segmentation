```python
import os
import shutil
import random
from tqdm import tqdm

random.seed(42)

# 원본 폴더
src_root = "data_aug"

# 출력 폴더
dst_root = "dataset"

train_img_dir = os.path.join(dst_root, "train", "Color_Images")
train_mask_dir = os.path.join(dst_root, "train", "Segmentation")

val_img_dir = os.path.join(dst_root, "val", "Color_Images")
val_mask_dir = os.path.join(dst_root, "val", "Segmentation")

os.makedirs(train_img_dir, exist_ok=True)
os.makedirs(train_mask_dir, exist_ok=True)
os.makedirs(val_img_dir, exist_ok=True)
os.makedirs(val_mask_dir, exist_ok=True)

all_samples = []

# -----------------------------------------------------------------------------
# 모든 데이터 수집
# -----------------------------------------------------------------------------

subfolders = sorted(os.listdir(src_root))

for folder in subfolders:

    folder_path = os.path.join(src_root, folder)

    if not os.path.isdir(folder_path):
        continue

    image_dir = os.path.join(folder_path, "image")
    mask_dir = os.path.join(folder_path, "mask")

    if not os.path.exists(image_dir) or not os.path.exists(mask_dir):
        continue

    image_files = sorted(os.listdir(image_dir))

    for img_name in image_files:

        img_path = os.path.join(image_dir, img_name)
        mask_path = os.path.join(mask_dir, img_name)

        if not os.path.exists(mask_path):
            continue

        all_samples.append((img_path, mask_path))

# -----------------------------------------------------------------------------
# 균일 셔플
# -----------------------------------------------------------------------------

random.shuffle(all_samples)

# -----------------------------------------------------------------------------
# train / val split
# -----------------------------------------------------------------------------

val_ratio = 0.2
n_total = len(all_samples)
n_val = int(n_total * val_ratio)

val_samples = all_samples[:n_val]
train_samples = all_samples[n_val:]

print(f"Total : {n_total}")
print(f"Train : {len(train_samples)}")
print(f"Val   : {len(val_samples)}")

# -----------------------------------------------------------------------------
# 저장 함수
# -----------------------------------------------------------------------------

def save_samples(samples, img_dst, mask_dst, start_idx=0):

    idx = start_idx

    for img_path, mask_path in tqdm(samples):

        new_name = f"{idx:07d}.png"

        shutil.copy(img_path, os.path.join(img_dst, new_name))
        shutil.copy(mask_path, os.path.join(mask_dst, new_name))

        idx += 1

    return idx

# -----------------------------------------------------------------------------
# 저장
# -----------------------------------------------------------------------------

idx = 0
idx = save_samples(train_samples, train_img_dir, train_mask_dir, idx)
idx = save_samples(val_samples, val_img_dir, val_mask_dir, idx)

print("Done!")
```
